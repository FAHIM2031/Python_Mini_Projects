```python
from cProfile import label

#Barchart
import matplotlib.pyplot as plt
product=['A','B','C','D','E','F']
sales=[1000,1500,800,1200,900,1000]
plt.bar(product,sales,color='orange',label='sales 2025')
plt.xlabel('product')
plt.ylabel('sales')
plt.title('product sales comparison')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
region=['North','South','West','East']
revenue=[3000,2000,1500,1200]
plt.pie(revenue,labels=region,autopct='%1.1f%%',colors=['blue','green','red','yellow'])
plt.title('revenue generated by Region')
plt.show()
```

```python
import matplotlib.pyplot as plt
hours_studies=[1,3,5,7,9,11,13,15]
exam_scores=[50,55,60,65,70,75,80,85]
plt.scatter(hours_studies,exam_scores,color='blue',label='exam',marker='>')
plt.scatter([2,4,6,8,10,12,14,16],[52,56,68,69,70,78,88,90],color='red',label='student',marker='o')
plt.xlabel('hours studied')
plt.ylabel('exam score')
plt.legend()
plt.title("Relationship between Study time and Exam score")
plt.grid(True)
plt.show()
```

```python
#plt.subplot(nrows,ncols,index)
import matplotlib.pyplot as plt
x=[1,2,3,4]
y=[5,6,7,8]
plt.subplot(1,2,1)
plt.bar(x,y)
plt.title('bar plot')
plt.subplot(1,2,2)
plt.pie(x,labels=y,autopct='%1.1f%%')
plt.title('pie chart')
plt.tight_layout()
plt.suptitle('Comparison of Line and Bar charts')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4]
y=[5,6,7,8]
fig,ax=plt.subplots(1,2,figsize=(10,5))
fig.suptitle('Comparison of Bar charts , Lines and Pie charts')
ax[0].plot(x,y)
ax[0].set_title('x axis')
ax[1].bar(x,y)
ax[1].set_title('y axis')

plt.savefig('all_subplots_e.png',dpi=300,bbox_inches='tight')

fig,ax1=plt.subplots(2,2,figsize=(10,5))
ax1[0,0].pie(x,labels=y,autopct='%1.1f%%')
ax1[0,0].set_title('pie Chart')

ax1[0,1].bar(x,y)
ax1[0,1].set_title('Bar Chart')
ax1[1,0].pie(x,labels=y,autopct='%1.1f%%')
ax1[1,0].set_title('pie Chart')

ax1[1,1].hist(x,bins=10,color='blue',edgecolor='white')
ax1[1,1].set_title('Histogram')
fig.suptitle('Comparison of Bar charts , Lines and Pie charts')
fig.tight_layout()
plt.savefig('all_subplots.png',dpi=300,bbox_inches='tight')

plt.show()
```
